```python
import geemap.geemap as geemap
import ee
```

```python
ee.Authenticate()
ee.Initialize(project = 'neon-feat-313910')
```

```python
base = geemap.Map()
```

```python
Districts = ee.FeatureCollection('projects/neon-feat-313910/assets/District_corrected')
Districts
```

```python
#Using LandSat7 2000 data
Bangalore = Districts.filter(ee.Filter.eq('District', 'BENGALORU_URBAN'))
base.addLayer(Bangalore, {}, 'Bangalore')
landsat_TCC_2000 = ee.ImageCollection(ee.ImageCollection("LANDSAT/LE07/C02/T1")).select('B4','B3','B2').filterDate('2000-04-01', '2000-05-15').filterMetadata('CLOUD_COVER', 'less_than', 10)
landsat_FCC_2000 = ee.ImageCollection(ee.ImageCollection("LANDSAT/LE07/C02/T1")).select('B5','B4','B3').filterDate('2000-04-01', '2000-05-15').filterMetadata('CLOUD_COVER', 'less_than', 10)
Bangalore_TCC_2000 = landsat_TCC_2000.mean().clip(Bangalore)
Bangalore_FCC_2000 = landsat_FCC_2000.mean().clip(Bangalore)
base.addLayer(Bangalore_TCC_2000, {}, 'Bangalore_TCC_2000')
base.addLayer(Bangalore_FCC_2000, {}, 'Bangalore_FCC_2000')
```

```python
# Using LandSat9 2024 data
landsat_TCC_2024 = ee.ImageCollection(ee.ImageCollection("LANDSAT/LC09/C02/T1_TOA")).select('B4','B3','B2').filterDate('2024-04-01', '2024-05-15').filterMetadata('CLOUD_COVER', 'less_than', 10)
landsat_FCC_2024 = ee.ImageCollection(ee.ImageCollection("LANDSAT/LC09/C02/T1_TOA")).select('B5','B4','B3').filterDate('2024-04-01', '2024-05-15').filterMetadata('CLOUD_COVER', 'less_than', 10)
Bangalore_TCC_2024 = landsat_TCC_2024.mean().clip(Bangalore)
Bangalore_FCC_2024 = landsat_FCC_2024.mean().clip(Bangalore)
base.addLayer(Bangalore_TCC_2024, {}, 'Bangalore_TCC_2024')
base.addLayer(Bangalore_FCC_2024, {}, 'Bangalore_FCC_2024')
```

```python
base.centerObject(Bangalore,9)
base
```
