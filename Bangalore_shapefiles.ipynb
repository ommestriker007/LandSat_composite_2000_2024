```python
import geemap.geemap as geemap
import ee
```

```python
ee.Authenticate()
ee.Initialize(project = 'neon-feat-313910')
```

```python
base = geemap.Map()
Bangalore = ee.FeatureCollection('projects/neon-feat-313910/assets/Bangalore')
Railways = ee.FeatureCollection('projects/neon-feat-313910/assets/railways')
base.addLayer(Bangalore, {}, 'Bangalore')
base.addLayer(Railways, {'color': 'red', 'width': 2, 'lineType': 'solid'}, 'Railways')
base.center_object(Bangalore,9)
base
```
