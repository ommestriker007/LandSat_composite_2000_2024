```python
import geemap.geemap as geemap
import ee
```

```python
ee.Authenticate()
ee.Initialize(project = 'neon-feat-313910')
```

```python
base = geemap.Map()
```

```python
Districts = ee.FeatureCollection('projects/neon-feat-313910/assets/District_corrected')
```

```python
# Using Sentinel 2016 data
Bangalore = Districts.filter(ee.Filter.eq('District', 'BENGALORU_URBAN'))
base.addLayer(Bangalore, {}, 'Bangalore')
Sentinel_TCC_2015 = ee.ImageCollection(ee.ImageCollection("COPERNICUS/S2_HARMONIZED")).select('B4','B3','B2').filterDate('2016-07-01', '2017-07-15').filterMetadata('CLOUD_COVER', 'less_than', 10)
Sentinel_FCC_2015 = ee.ImageCollection(ee.ImageCollection("COPERNICUS/S2_HARMONIZED")).select('B8','B4','B3').filterDate('2016-07-01', '2017-07-15').filterMetadata('CLOUD_COVER', 'less_than', 10)
Bangalore_TCC_2015 = Sentinel_TCC_2015.mean().clip(Bangalore)
Bangalore_FCC_2015 = Sentinel_FCC_2015.mean().clip(Bangalore)
base.addLayer(Sentinel_TCC_2015, {}, 'Bangalore_TCC_2015')
base.addLayer(Sentinel_FCC_2015, {}, 'Bangalore_FCC_2015')
```

```python
# Using Sentinel 2024 data
Sentinel_TCC_2024 = ee.ImageCollection(ee.ImageCollection("COPERNICUS/S2_HARMONIZED")).select('B4','B3','B2').filterDate('2024-01-01', '2024-07-15').filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20))
Sentinel_FCC_2024 = ee.ImageCollection(ee.ImageCollection("COPERNICUS/S2_HARMONIZED")).select('B8','B4','B3').filterDate('2024-01-01', '2024-07-15').filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20))
Bangalore_TCC_2024 = Sentinel_TCC_2024.mean().clip(Bangalore)
Bangalore_FCC_2024 = Sentinel_FCC_2024.mean().clip(Bangalore)
base.addLayer(Sentinel_TCC_2024, {}, 'Bangalore_TCC_2024')
base.addLayer(Sentinel_FCC_2024, {}, 'Bangalore_FCC_2024')
```
